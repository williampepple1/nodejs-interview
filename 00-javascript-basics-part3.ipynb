## Section 9: Arrays (MCQs 101-120)

Arrays are ordered, indexed collections. JavaScript provides many built-in methods for manipulation, iteration, and transformation. Key concepts: mutating vs non-mutating methods, iteration (map, forEach, filter, reduce), and utility methods like slice, splice, find, and includes.

**MCQ 101:** Which method adds elements to the end of an array?

A) unshift  
B) push  
C) pop  
D) shift

<details>
<summary>Answer</summary>

**B) push** — `push()` adds elements to the end of an array and returns the new length.
</details>

**MCQ 102:** Which method removes the last element?

A) shift  
B) pop  
C) splice  
D) slice

<details>
<summary>Answer</summary>

**B) pop** — `pop()` removes and returns the last element of an array.
</details>

**MCQ 103:** What is the difference between `map()` and `forEach()`?

A) No difference  
B) map returns a new array, forEach returns undefined  
C) forEach returns new array  
D) map modifies original

<details>
<summary>Answer</summary>

**B)** — `map()` returns a new array with transformed elements; `forEach()` returns `undefined` and is used for side effects only.
</details>

**MCQ 104:** What is the output of `[1,2,3].map(x => x * 2)`?

A) [1,2,3]  
B) [2,4,6]  
C) 12  
D) undefined

<details>
<summary>Answer</summary>

**B) [2,4,6]** — `map()` applies the callback to each element and returns a new array with the results.
</details>

```javascript
// MCQ 104 - Verify output
console.log([1,2,3].map(x => x * 2));
```

**MCQ 105:** What does `filter()` do?

A) Sorts array  
B) Returns new array with elements that pass the test function  
C) Removes duplicates  
D) Flattens array

<details>
<summary>Answer</summary>

**B)** — `filter()` returns a new array containing only elements for which the callback returns a truthy value.
</details>

**MCQ 106:** What does `reduce()` do?

A) Reduces array size  
B) Executes a reducer function on each element, accumulating a single result  
C) Filters values  
D) Sorts values

<details>
<summary>Answer</summary>

**B)** — `reduce()` runs a reducer on each element, accumulating a single value. Example: `[1,2,3].reduce((acc, x) => acc + x, 0)` = 6.
</details>

**MCQ 107:** What is the output of `[1,2,3].reduce((acc, x) => acc + x, 0)`?

A) [1,2,3]  
B) 6  
C) "123"  
D) 0

<details>
<summary>Answer</summary>

**B) 6** — The reducer sums all elements: 0+1=1, 1+2=3, 3+3=6.
</details>

```javascript
// MCQ 107 - Verify output
console.log([1,2,3].reduce((acc, x) => acc + x, 0));
```

**MCQ 108:** What is the difference between `slice()` and `splice()`?

A) No difference  
B) slice returns a copy without modifying original, splice modifies original  
C) splice returns a copy  
D) slice modifies original

<details>
<summary>Answer</summary>

**B)** — `slice(start, end)` extracts a portion and returns a new array (end exclusive). `splice(start, deleteCount, ...items)` mutates the original array.
</details>

**MCQ 109:** What is the output of `[1,2,3,4,5].slice(1,3)`?

A) [1,2,3]  
B) [2,3]  
C) [2,3,4]  
D) [1,2]

<details>
<summary>Answer</summary>

**B) [2,3]** — `slice(1,3)` returns elements at index 1 and 2 (end is exclusive).
</details>

```javascript
// MCQ 109 - Verify output
console.log([1,2,3,4,5].slice(1,3));
```

**MCQ 110:** What does `find()` return?

A) An array of matches  
B) The first element that satisfies the test, or undefined  
C) The index  
D) Boolean

<details>
<summary>Answer</summary>

**B)** — `find()` returns the first element for which the callback returns truthy, or `undefined` if none.
</details>

**MCQ 111:** What does `findIndex()` return?

A) The element  
B) The index of the first element that passes test, or -1  
C) An array  
D) Boolean

<details>
<summary>Answer</summary>

**B)** — `findIndex()` returns the index of the first matching element, or -1 if none.
</details>

**MCQ 112:** What does `some()` vs `every()` do?

A) Same thing  
B) some returns true if ANY element passes; every returns true if ALL pass  
C) some checks first, every checks last  
D) some is async

<details>
<summary>Answer</summary>

**B)** — `some()` returns true if at least one element passes; `every()` returns true only if all elements pass.
</details>

**MCQ 113:** What does `flat()` do?

A) Sorts array  
B) Flattens nested arrays: `[1,[2,[3]]].flat(Infinity)` → `[1,2,3]`  
C) Removes duplicates  
D) Reverses array

<details>
<summary>Answer</summary>

**B)** — `flat(depth)` flattens nested arrays. Use `flat(Infinity)` for full flattening.
</details>

**MCQ 114:** How to remove duplicates from an array?

A) `arr.unique()`  
B) `[...new Set(arr)]`  
C) `arr.distinct()`  
D) `arr.filter()` only

<details>
<summary>Answer</summary>

**B)** — `Set` stores unique values only. Spread converts it back to an array.
</details>

**MCQ 115:** What does `Array.from()` do?

A) Deletes an array  
B) Creates a new array from an iterable or array-like object  
C) Checks if value is array  
D) Converts to string

<details>
<summary>Answer</summary>

**B)** — `Array.from('hello')` → ['h','e','l','l','o']. `Array.from({length:3}, (_, i) => i)` → [0,1,2].
</details>

**MCQ 116:** What is the output of `[10, 9, 8].sort()`?

B) [10, 8, 9]

D) [1, 8, 9, 10]

<details>

**B) [10, 8, 9]** — Default `sort()` converts elements to strings! "10" < "8" (character comparison). Fix: `.sort((a,b) => a-b)`.
</details>

```javascript
// MCQ 116 - Verify output
console.log([10, 9, 8].sort());
```

**MCQ 120:** What is array destructuring?

## Section 10: Objects (MCQs 121-135)

Objects are key-value collections. Key concepts: creation (literal, constructor, Object.create), property access (dot vs bracket), utility methods (keys, values, entries, assign), immutability (freeze), copying (shallow vs deep), and destructuring.

**MCQ 121:** How to create an object?

A) Only with classes  
B) Object literal `{}`, `new Object()`, `Object.create()`, class constructor  
C) Only `new Object()`  
D) Only object literal

<details>
<summary>Answer</summary>

**B)** — Multiple ways: `{}`, `new Object()`, `Object.create(proto)`, or via class constructor.
</details>

**MCQ 122:** What is the difference between dot notation and bracket notation?

A) No difference  
B) Bracket notation allows dynamic keys and keys with special characters; dot notation only works with valid identifiers  
C) Dot is faster  
D) Bracket is deprecated

<details>
<summary>Answer</summary>

**B)** — `obj[key]` works with variables and keys like `'my-key'`; `obj.key` requires valid identifier names.
</details>

**MCQ 123:** What does `Object.keys()` return?

A) Object values  
B) An array of the object's own enumerable property names  
C) Prototype properties  
D) All properties including inherited

<details>
<summary>Answer</summary>

**B)** — Returns only own enumerable property keys, not inherited ones.
</details>

**MCQ 124:** What does `Object.values()` return?

A) Property names  
B) An array of the object's own enumerable property values  
C) Methods only  
D) Key-value pairs

<details>
<summary>Answer</summary>

**B)** — Returns an array of values for own enumerable properties.
</details>

**MCQ 125:** What does `Object.entries()` return?

A) Keys only  
B) An array of [key, value] pairs  
C) Values only  
D) A Map

<details>
<summary>Answer</summary>

**B)** — `Object.entries({a:1,b:2})` → [['a',1],['b',2]]. Useful with `for...of` and `Object.fromEntries()`.
</details>

**MCQ 126:** What does `Object.assign()` do?

A) Compares objects  
B) Copies properties from source objects to a target object (shallow copy)  
C) Deep clones  
D) Deletes properties

<details>
<summary>Answer</summary>

**B)** — `Object.assign({}, obj1, obj2)`. Only copies own enumerable properties. Shallow copy only!
</details>

**MCQ 127:** What does `Object.freeze()` do?

A) Deletes the object  
B) Makes the object immutable — can't add, remove, or modify properties (shallow freeze)  
C) Copies the object  
D) Converts to array

<details>
<summary>Answer</summary>

**B)** — Only shallow — nested objects are still mutable. Use libraries for deep freeze.
</details>

**MCQ 128:** Difference between shallow copy and deep copy?

A) Same thing  
B) Shallow copy copies references for nested objects; deep copy creates completely independent copies  
C) Shallow is faster  
D) Deep copy is impossible in JS

<details>
<summary>Answer</summary>

**B)** — Shallow: `{...obj}`, `Object.assign()`. Deep: `structuredClone(obj)` (modern), `JSON.parse(JSON.stringify(obj))` (limited — no functions, dates, undefined).
</details>

B) 2

**B)** — Two special differences: NaN equals NaN, and +0 does not equal -0.

## Section 11: Prototypes & Inheritance (MCQs 136-143)

JavaScript uses prototypal inheritance: objects inherit from other objects via the prototype chain. Key concepts: `__proto__` vs `prototype`, `Object.create()`, `hasOwnProperty()`, `instanceof`, and `Object.getPrototypeOf()`.

**MCQ 136:** What is the prototype chain?

A) A linked list of arrays  
B) Every object has an internal prototype link; when a property is not found, JavaScript walks up the chain to the prototype  
C) CSS inheritance  
D) Module dependencies

<details>
<summary>Answer</summary>

**B)** — Objects delegate to their prototype when a property isn't found on the object itself.
</details>

**MCQ 137:** What does `Object.create()` do?

A) Copies an object  
B) Creates a new object with the specified object as its prototype  
C) Freezes an object  
D) Deletes an object

<details>
<summary>Answer</summary>

**B)** — `const child = Object.create(parent)` — `child` inherits from `parent`.
</details>

**MCQ 138:** What is `__proto__` vs `prototype`?

A) Same thing  
B) `__proto__` is the actual prototype link on every object instance; `prototype` is a property on constructor functions used to set `__proto__` of created instances  
C) `__proto__` is deprecated only  
D) `prototype` is on instances

<details>
<summary>Answer</summary>

**B)** — `__proto__` is on instances (the link); `prototype` is on constructors (the object that becomes `__proto__` of instances).
</details>

**MCQ 139:** What does `hasOwnProperty()` do?

A) Checks prototype  
B) Returns true if the property is directly on the object (not inherited from prototype)  
C) Adds a property  
D) Deletes a property

<details>
<summary>Answer</summary>

**B)** — Checks only own properties, not those from the prototype chain.
</details>

**MCQ 140:** Output: `console.log([1,2,3].hasOwnProperty('length'))`

A) true  
B) false  
C) TypeError  
D) undefined

<details>
<summary>Answer</summary>

**A) true** — `length` is an own property of arrays, not inherited.
</details>

```javascript
// MCQ 140 - Verify output
console.log([1,2,3].hasOwnProperty('length'));
```

**MCQ 141:** How does `instanceof` work?

A) Checks value equality  
B) Checks if an object's prototype chain includes the constructor's prototype  
C) Checks type of primitive  
D) Compares strings

<details>
<summary>Answer</summary>

**B)** — `[] instanceof Array` → true. `[] instanceof Object` → true (Array inherits from Object).
</details>

**MCQ 142:** What does `Object.getPrototypeOf(obj)` return?

A) The object itself  
B) The prototype of the object  
C) The constructor  
D) undefined

<details>
<summary>Answer</summary>

**B)** — Preferred over `obj.__proto__` (which is deprecated).
</details>

**MCQ 143:** What is prototypal inheritance vs classical inheritance?

A) Same thing  
B) Prototypal: objects inherit directly from other objects. Classical: classes inherit from classes (JS simulates this with prototypes under the hood)  
C) Prototypal uses extends  
D) Classical is JS native

<details>
<summary>Answer</summary>

**B)** — JavaScript is prototypal; ES6 classes provide a classical-style syntax built on prototypes.
</details>

## Section 12: Classes (MCQs 144-150)

ES6 classes are syntactic sugar over prototypal inheritance. Key concepts: constructor, extends, super, static methods, getters/setters, and private fields (`#`).

**MCQ 144:** What are ES6 classes in JavaScript?

A) A new type of object  
B) Syntactic sugar over prototypal inheritance  
C) Completely different from prototypes  
D) Compiled to different bytecode

<details>
<summary>Answer</summary>

**B)** — Classes are syntactic sugar — they still use prototypes under the hood.
</details>

**MCQ 145:** What is the `constructor` method in a class?

A) A static method  
B) A special method for initializing objects created with `new`  
C) A getter  
D) An optional decorator

<details>
<summary>Answer</summary>

**B)** — Called automatically when creating instances with `new`.
</details>

**MCQ 146:** What is `extends` used for?

A) Adding properties  
B) Creating a class that inherits from another class  
C) Exporting  
D) Importing

<details>
<summary>Answer</summary>

**B)** — `class Dog extends Animal { }`. The child class inherits all methods from the parent.
</details>

**MCQ 147:** What does `super` do?

A) Makes a function async  
B) Calls the parent class constructor or methods  
C) Creates a global variable  
D) Returns `this`

<details>
<summary>Answer</summary>

**B)** — `super()` in constructor calls parent constructor. `super.method()` calls parent method. Must call `super()` before using `this` in a derived constructor.
</details>

**MCQ 148:** What are static methods?

A) Methods that can't change  
B) Methods called on the class itself, not on instances: `static create() {}` → `MyClass.create()`  
C) Private methods  
D) Async methods

<details>
<summary>Answer</summary>

**B)** — `Array.isArray()`, `Object.keys()`, `Number.isNaN()` are static methods.
</details>

**MCQ 149:** What are getters and setters?

A) HTTP methods  
B) Special methods that let you define property access and assignment behavior: `get name() {}` and `set name(val) {}`  
C) Array methods  
D) Module exports

<details>
<summary>Answer</summary>

**B)** — Customize property read/write behavior.
</details>

**MCQ 150:** What is `#privateField` in classes?

A) A comment  
B) A truly private class field (ES2022) that can only be accessed inside the class  
C) A CSS selector  
D) A template literal

<details>
<summary>Answer</summary>

**B)** — `class Foo { #secret = 42; getSecret() { return this.#secret; } }`. Accessing `foo.#secret` from outside throws SyntaxError.
</details>
