## 50 Detailed Interview Q&As

### Q1: Explain the difference between `var`, `let`, and `const` with examples.

**Model Answer:**

`var` is function-scoped, hoisted (initialized as undefined), and allows redeclaration. `let` is block-scoped, hoisted but in the temporal dead zone (ReferenceError if accessed before declaration), and does not allow redeclaration. `const` is like `let` but cannot be reassigned (though object properties can be mutated). Best practice: prefer `const`, use `let` when reassignment is needed, avoid `var`.

```javascript
var x = 1; var x = 2;  // OK
let y = 1; let y = 2;  // SyntaxError
const z = 1; z = 2;    // TypeError
const obj = {}; obj.a = 1;  // OK - mutating property
```

### Q2: What is hoisting? Give examples with var, let, functions.

**Model Answer:**

Hoisting is JS behavior of moving declarations to the top of their scope during compilation. `var` is hoisted and initialized as undefined. `let`/`const` are hoisted but NOT initialized (temporal dead zone). Function declarations are fully hoisted (can call before definition). Function expressions follow variable hoisting rules.

```javascript
console.log(a);  // undefined (var hoisted)
var a = 1;
console.log(b);  // ReferenceError (let in TDZ)
let b = 2;
foo();  // works - function declaration hoisted
function foo() {}
```

### Q3: Explain type coercion in JavaScript with examples.

**Model Answer:**

JavaScript automatically converts types during operations. String coercion: `'5' + 3 = '53'` (+ with string concatenates). Numeric coercion: `'5' - 3 = 2` (-, *, / force numbers). Boolean coercion: `if('hello')` (truthy). `==` does coercion, `===` doesn't. Explicit conversion: `Number('5')`, `String(5)`, `Boolean(0)`.

### Q4: List all falsy values in JavaScript.

**Model Answer:**

8 falsy values: `false`, `0`, `-0`, `0n` (BigInt zero), `""` (empty string), `null`, `undefined`, `NaN`. Everything else is truthy, including `[]`, `{}`, `"0"`, `"false"`, `new Boolean(false)`.

### Q5: Explain closures with a practical example.

**Model Answer:**

A closure is when a function retains access to its outer scope's variables even after the outer function returns.

```javascript
function counter() {
  let count = 0;
  return () => ++count;
}
const inc = counter();
inc(); // 1
inc(); // 2
```

Practical uses: data privacy, module pattern, event handlers, memoization, function factories.

### Q6: What is the difference between `==` and `===`?

**Model Answer:**

`==` (loose equality) performs type coercion before comparison. `===` (strict equality) checks type AND value without coercion. Examples: `'5' == 5` is true, `'5' === 5` is false. `null == undefined` is true, `null === undefined` is false. Always use `===`.

### Q7: Explain the `this` keyword with examples in different contexts.

**Model Answer:**

Global context: `this` = global object (or undefined in strict mode). Object method: `this` = the calling object. Arrow function: `this` = inherited from lexical scope. Constructor (with new): `this` = the new object. call/apply/bind: `this` = explicitly set. Event handler: `this` = the element (unless arrow function).

### Q8: What is the prototype chain?

**Model Answer:**

Every JS object has an internal [[Prototype]] link. When accessing a property, JS first checks the object itself, then walks up the prototype chain. Arrays → Array.prototype → Object.prototype → null. This is how inheritance works in JS.

### Q9: What are Promises and how do they work?

**Model Answer:**

A Promise represents an eventual async result. States: pending → fulfilled (resolve) or rejected (reject). Consume with .then()/.catch()/.finally(). Chain for sequential async ops. Combinators: Promise.all(), allSettled(), race(), any(). Always handle rejections.

### Q10: Explain async/await.

**Model Answer:**

Syntactic sugar over Promises. async function always returns a Promise. await pauses async function until Promise settles. Use try/catch for error handling. Makes async code look synchronous. await only pauses the function, not the program.

### Q11: What is the event loop?

**Model Answer:**

Mechanism that handles async operations in single-threaded JS. Call stack executes synchronous code. When stack is empty, event loop checks microtask queue (Promises, nextTick), then macrotask queue (setTimeout, I/O). Microtasks are processed completely before each macrotask.

### Q12: Explain `map`, `filter`, and `reduce`.

**Model Answer:**

map: transforms each element, returns new array. `[1,2,3].map(x => x*2)` → [2,4,6]. filter: returns new array with elements passing test. `[1,2,3].filter(x => x>1)` → [2,3]. reduce: accumulates single value. `[1,2,3].reduce((acc,x) => acc+x, 0)` → 6. None mutate the original array.

### Q13: What is destructuring? Give examples for arrays and objects.

**Model Answer:**

Extracting values into variables. Array: `const [a, b, ...rest] = [1, 2, 3, 4]`. Object: `const { name, age: years } = person`. Defaults: `const { x = 10 } = {}`. Nested: `const { address: { city } } = person`. Swap: `[a, b] = [b, a]`.

### Q14: Explain the spread and rest operators.

**Model Answer:**

Same syntax `...` but different contexts. Spread (expands): `[...arr]` copies array, `{...obj}` copies object, `fn(...args)` spreads arguments. Rest (collects): `function(...args)` collects into array, `const [first, ...rest] = arr` collects remaining.

### Q15: What is the difference between `null` and `undefined`?

**Model Answer:**

undefined: variable declared but not assigned, missing function parameters, missing object properties, functions with no return. null: explicitly assigned by developer to indicate intentional absence. typeof undefined is "undefined", typeof null is "object" (bug). null == undefined is true, null === undefined is false.

### Q16: How do arrow functions differ from regular functions?

**Model Answer:**

No own `this` (inherits lexically), no `arguments` object, can't be used as constructors (no `new`), no `prototype` property, implicit return for single expressions. Best for callbacks and methods that don't need their own `this`.

### Q17: Explain `call()`, `apply()`, and `bind()`.

**Model Answer:**

All set `this` explicitly. call(obj, arg1, arg2): invokes immediately with individual args. apply(obj, [args]): invokes immediately with args array. bind(obj): returns new function with `this` permanently bound. Use case: borrowing methods from other objects.

### Q18: What is the module pattern?

**Model Answer:**

Uses IIFE + closures to create private state with public API.

```javascript
const mod = (function() {
  let priv = 0;
  return { get: () => priv, inc: () => ++priv };
})();
```

Precursor to ES modules. Still useful for encapsulation.

### Q19: Explain ES modules vs CommonJS.

**Model Answer:**

ES modules: import/export, static analysis, tree-shaking, async loading, strict mode default. CommonJS: require/module.exports, dynamic, synchronous, no tree-shaking. ESM is the standard; CJS is Node.js legacy. Use `"type":"module"` in package.json for ESM.

### Q20: What are generator functions?

**Model Answer:**

Functions that can pause/resume with `yield`. Declared with `function*`. Return iterator with `.next()`. Produce `{value, done}`. Use cases: lazy evaluation, infinite sequences, custom iterables, async control flow (before async/await).

### Q21: Explain `Object.freeze()` vs `Object.seal()` vs `Object.preventExtensions()`.

**Model Answer:**

preventExtensions: can't add new properties (can modify/delete existing). seal: can't add/delete properties (can modify existing values). freeze: can't add/delete/modify any properties (most restrictive). All are shallow — nested objects aren't affected.

### Q22: What is memoization?

**Model Answer:**

Caching function results for given arguments. Avoids recomputation. Implementation: closure with cache object, key = serialized arguments. Great for expensive computations with repeated inputs.

```javascript
const memo = fn => {
  const c = {};
  return (...a) => c[JSON.stringify(a)] ?? (c[JSON.stringify(a)] = fn(...a));
};
```

### Q23: Explain event bubbling and capturing.

**Model Answer:**

When event occurs on nested element: Capturing phase goes root → target. Target phase fires on the element. Bubbling phase goes target → root (default behavior). `addEventListener(event, handler, {capture: true})` for capture phase. `event.stopPropagation()` prevents further propagation.

### Q24: What is event delegation?

**Model Answer:**

Instead of attaching listeners to each child, attach one listener to the parent. Use `event.target` to identify which child triggered it. Benefits: works with dynamically added children, less memory, fewer listeners.

```javascript
parent.addEventListener('click', e => {
  if (e.target.matches('.btn')) handleClick(e.target);
});
```

### Q25: What is debouncing and throttling?

**Model Answer:**

Debounce: delay execution until N ms after last call (search input — wait for user to stop typing). Throttle: execute at most once per N ms (scroll/resize handlers). Implementations use setTimeout and closures. Libraries: lodash.debounce, lodash.throttle.

### Q26: Explain `Map` and `Set` with use cases.

**Model Answer:**

Map: key-value pairs where keys can be ANY type (not just strings). Has .size, .get(), .set(), .has(), .delete(). Use: object keys, caching. Set: unique values only. Has .size, .add(), .has(), .delete(). Use: removing duplicates `[...new Set(arr)]`, membership checks.

### Q27: What are WeakMap and WeakSet?

**Model Answer:**

Keys/values must be objects. Entries are garbage-collected when no other reference exists (weak references). WeakMap: private data storage, DOM node metadata. WeakSet: tracking objects without preventing GC. Neither is iterable and neither has .size.

### Q28: Explain the difference between shallow copy and deep copy.

**Model Answer:**

Shallow: copies top-level properties, nested objects share references. Methods: spread `{...obj}`, Object.assign(), `[...arr]`. Deep: creates fully independent copies including nested. Methods: structuredClone() (modern), JSON.parse(JSON.stringify()) (loses functions/dates/undefined). Nested mutation test reveals the difference.

### Q29: What is currying and partial application?

**Model Answer:**

Currying: `f(a,b,c)` → `f(a)(b)(c)`. Each call returns a new function taking next arg. Partial application: fixing some arguments, returning function for the rest. `const add5 = add.bind(null, 5)`. Use cases: configuration, function composition, reusable utilities.

### Q30: Explain `Symbol` and its use cases.

**Model Answer:**

Unique, immutable primitive. `Symbol('desc')` — desc is just a label. Use cases: unique object property keys (no name collision), well-known symbols (Symbol.iterator, Symbol.toPrimitive), private-ish properties (not truly private — Object.getOwnPropertySymbols reveals them).

### Q31: What is optional chaining and nullish coalescing?

**Model Answer:**

Optional chaining `?.`: access nested properties safely. `user?.address?.city` returns undefined if any intermediate is null/undefined. Works with methods: `obj.fn?.()`. Nullish coalescing `??`: fallback only for null/undefined. `value ?? 'default'`. Unlike `||`, doesn't trigger on 0, '', false.

### Q32: What is the difference between `for...in` and `for...of`?

**Model Answer:**

`for...in`: iterates over enumerable property KEYS (strings), including inherited. Best for objects. `for...of`: iterates over iterable VALUES (array elements, string characters, Map entries). Best for arrays/iterables. Never use for...in on arrays (gets index strings, includes prototype properties).

### Q33: Explain `try...catch...finally`.

**Model Answer:**

try: code that might throw. catch(e): handles the error (e = the thrown value). finally: always runs regardless of error — for cleanup (close files, release resources). finally runs even if try/catch has return. Can nest try/catch. Async errors need await inside try.

### Q34: What is the difference between `throw new Error()` and `throw 'string'`?

**Model Answer:**

Error objects include: message, name, stack trace (crucial for debugging). Throwing strings loses the stack trace. Custom errors: extend Error class. Always throw Error objects. catch block should check error type: `if (e instanceof TypeError)`.

### Q35: What is `Object.keys()` vs `Object.getOwnPropertyNames()` vs `Reflect.ownKeys()`?

**Model Answer:**

Object.keys(): own enumerable string keys. Object.getOwnPropertyNames(): own string keys (enumerable + non-enumerable). Reflect.ownKeys(): ALL own keys (strings + symbols, enumerable + non-enumerable). Object.entries() gives [key, value] pairs for own enumerable.

### Q36: Explain template literals and tagged templates.

**Model Answer:**

Template literals: backtick strings with `${expression}` interpolation and multi-line support. Tagged templates: `tag\`Hello ${name}\`` — tag function receives string parts array and values. Use cases: SQL query builders, CSS-in-JS (styled-components), i18n, sanitization.

### Q37: What is the Iterator protocol?

**Model Answer:**

An object is iterable if it has `[Symbol.iterator]()` that returns an iterator. An iterator has a `next()` method returning `{value, done}`. Built-in iterables: Array, String, Map, Set, TypedArray, arguments. Consumed by: for...of, spread, destructuring, Array.from().

### Q38: What is a Proxy and Reflect?

**Model Answer:**

Proxy wraps an object and intercepts operations via traps (get, set, has, deleteProperty, apply, construct, etc.). Reflect provides default behavior for each trap. Use cases: validation, logging, computed properties, ORMs, reactive frameworks (Vue 3).

### Q39: Explain `structuredClone()`.

**Model Answer:**

Modern deep cloning API. Handles: nested objects, arrays, Maps, Sets, Dates, RegExps, ArrayBuffers, circular references. Doesn't handle: functions, DOM nodes, property descriptors, getters/setters. Better than JSON method (which fails on dates, undefined, circular refs, Maps, Sets).

### Q40: What is the temporal dead zone (TDZ)? Why does it exist?

**Model Answer:**

Period between scope entry and variable declaration for let/const. Accessing throws ReferenceError. Exists to catch bugs — accessing variables before declaration is almost always a mistake. var's hoisting to undefined masks these bugs. TDZ makes code more predictable.

### Q41: Explain `Object.defineProperty()`.

**Model Answer:**

Defines/modifies a property with fine control. Descriptors: `value`, `writable` (can change?), `enumerable` (shows in for...in?), `configurable` (can delete/modify descriptor?). Also: `get`/`set` for accessor properties. Used by frameworks for reactivity (Vue 2).

### Q42: What is the difference between `Array.from()` and spread operator for converting iterables?

**Model Answer:**

Both convert iterables to arrays. `Array.from()` also works with array-like objects (has length, indexed, but no array methods — like arguments, NodeList). `Array.from()` accepts a mapping function as second arg: `Array.from({length:3}, (_,i) => i)` → [0,1,2]. Spread requires actual iterator protocol.

### Q43: Explain short-circuit evaluation and its practical uses.

**Model Answer:**

`&&`: returns first falsy or last value. `||`: returns first truthy or last value. Uses: default values `const name = input || 'default'`, conditional execution `user && user.save()`, guard clauses `isValid && doSomething()`. Prefer `??` for defaults (handles 0 and '' correctly).

### Q44: What is a pure function and why does it matter?

**Model Answer:**

Same inputs → same output, no side effects. Benefits: predictable, testable, cacheable (memoizable), safe for concurrency. Impure: reads/writes globals, DOM, network, random, Date.now(). Functional programming principle. React components ideally are pure functions of props.

### Q45: Explain `this` in 5 different scenarios.

**Model Answer:**

1) Global: `this` = window/global. 2) Object method: `this` = calling object. 3) Arrow function: `this` = enclosing lexical `this`. 4) Constructor: `this` = new instance. 5) Explicit: call/apply/bind set `this`. Priority: new > explicit > method > default. Arrow functions ignore all binding.

### Q46: What are getters and setters?

**Model Answer:**

Special methods that intercept property access/assignment. Defined with `get prop()` and `set prop(val)`. Use cases: computed properties (fullName from first+last), validation (age must be positive), lazy loading, logging access. In classes or Object.defineProperty.

### Q47: Explain error handling best practices in JavaScript.

**Model Answer:**

Always throw Error objects (not strings). Use specific error types (TypeError, RangeError, custom). Catch at boundaries (API handlers, event listeners). Don't catch and ignore. Log with stack trace. Use finally for cleanup. Async: try/catch with await, or .catch(). Global: window.onerror, process.on('uncaughtException').

### Q48: What is the difference between classical inheritance and prototypal inheritance?

**Model Answer:**

Classical (Java, C++): classes are blueprints, instances copy from class. Prototypal (JavaScript): objects inherit directly from other objects via prototype chain. JS classes are syntactic sugar over prototypes. Favor composition over inheritance. Prototype chain: obj → Constructor.prototype → Object.prototype → null.

### Q49: Explain immutability in JavaScript.

**Model Answer:**

Preventing data mutation. Primitives are immutable by nature. Objects/arrays are mutable. Techniques: Object.freeze() (shallow), spread copies `{...obj, updated: true}`, array methods that return new arrays (map, filter, slice vs push, splice, sort). Libraries: Immutable.js, Immer. Benefits: predictable state, easier debugging, enables === comparisons for change detection.

**Model Answer:**

let/const (block scope), arrow functions (lexical this), template literals, destructuring, spread/rest, default params, Promises/async-await, classes, modules (import/export), Map/Set, Symbol, for...of, generators, optional chaining (?.), nullish coalescing (??), private class fields (#), structuredClone(), Array.at(), Object.hasOwn(). Each solved specific pain points from ES5 era.
