## Section 2: Event Loop MCQs

### MCQ 9
**What is the correct order of the Event Loop phases?**

A) poll → check → timers → pending callbacks → close callbacks  
B) timers → pending callbacks → idle/prepare → poll → check → close callbacks  
C) timers → poll → check → pending callbacks → close callbacks  
D) check → timers → poll → close callbacks → pending callbacks  

<details>
<summary><b>Click to reveal answer</b></summary>

**Answer: B) timers → pending callbacks → idle/prepare → poll → check → close callbacks**

The event loop phases in order: (1) **timers** — setTimeout/setInterval, (2) **pending callbacks** — deferred I/O callbacks, (3) **idle/prepare** — internal use, (4) **poll** — retrieve new I/O events, (5) **check** — setImmediate, (6) **close callbacks** — socket.on('close').
</details>

### MCQ 13
**Which statement about `process.nextTick()` is TRUE?**

A) It schedules a callback in the timers phase of the event loop  
B) It fires after `setImmediate()` but before `setTimeout()`  
C) It can starve the event loop if called recursively  
D) It is identical in behavior to `setImmediate()`  

<details>
<summary><b>Click to reveal answer</b></summary>

**Answer: C) It can starve the event loop if called recursively**

If you call `process.nextTick()` recursively, the nextTick queue never empties, so the event loop never advances to I/O, timers, or any other phase. This is a real production risk. Use `setImmediate()` for safe recursion that allows I/O between iterations.
</details>

### MCQ 14
**Inside an I/O callback, what is the order of `setTimeout(fn, 0)` vs `setImmediate(fn)`?**

A) `setTimeout` always fires first  
B) `setImmediate` always fires first  
C) They always fire at the same time  
D) The order is random and unpredictable  

<details>
<summary><b>Click to reveal answer</b></summary>

**Answer: B) `setImmediate` always fires first**

Inside an I/O callback, the event loop is in the **poll** phase. After poll, the next phase is **check** (where `setImmediate` runs), followed by close callbacks, and then timers on the next loop iteration. So `setImmediate` is deterministically first. Outside I/O, the order is non-deterministic.
</details>

### MCQ 15
**What is the relationship between microtasks and macrotasks?**

A) Macrotasks always run before microtasks  
B) Microtasks and macrotasks alternate one-for-one  
C) The entire microtask queue is drained between each macrotask  
D) There is no difference between microtasks and macrotasks  

<details>
<summary><b>Click to reveal answer</b></summary>

**Answer: C) The entire microtask queue is drained between each macrotask**

After each phase of the event loop (each macrotask), Node.js checks and drains ALL pending microtasks (nextTick queue first, then Promise queue) before proceeding to the next phase/macrotask. This is why microtasks added during microtask processing also run before any macrotask.
</details>

## Fundamentals — Long-Form Questions

### Q1: Explain the complete architecture of Node.js.

**Model Answer:**

Node.js has a layered architecture:

1. **Application layer** — Your JavaScript code (Express routes, business logic, etc.)

2. **Node.js Bindings** — A C++ bridge layer that connects JavaScript to lower-level C/C++ libraries. This allows JS code to call native OS operations.

3. **V8 Engine** — Google's open-source JavaScript engine that compiles JS to machine code using JIT compilation (Ignition interpreter → TurboFan optimizing compiler).

4. **libuv** — A C library that provides:
   - The **event loop** (the heart of Node.js async behavior)
   - A **thread pool** (default 4 threads) for blocking operations
   - Asynchronous wrappers around OS-level I/O (file system, networking, DNS)

5. **Operating System** — The lowest layer. Network I/O uses OS async primitives (epoll on Linux, kqueue on macOS, IOCP on Windows).

The key insight is that JavaScript is single-threaded, but Node.js as a whole is NOT — libuv uses multiple threads behind the scenes for operations that can't be done asynchronously at the OS level.

### Q2: Walk me through the complete lifecycle of an HTTP request in Node.js.

**Model Answer:**

1. **Client sends request** — TCP connection established via OS networking
2. **libuv's poll phase** picks up the new connection event
3. The **connection callback** fires on the main thread (your `http.createServer` callback)
4. **Middleware/routing** — Express processes the request through the middleware pipeline
5. **Request handler** runs your business logic:
   - If it's **synchronous** (computation) → blocks the event loop
   - If it's **async I/O** (DB query, file read) → delegates to libuv, main thread is free
6. **Async result arrives** — libuv queues the callback
7. **Event loop** picks up the callback, executes it on the main thread
8. **Response sent** — `res.json()` or `res.send()` writes to the TCP socket
9. **Socket closed** or kept alive (HTTP keep-alive)

The critical takeaway: during step 5, if the operation is async, the main thread is free to handle OTHER requests. This is why Node.js can handle thousands of concurrent connections on a single thread.

### Q3: Explain the Event Loop in detail. What are microtasks and macrotasks?

**Model Answer:**

The event loop is the mechanism that allows Node.js to perform non-blocking I/O despite being single-threaded. It processes work in **phases**:

**Phases (macrotask queues):**
1. **Timers** — Execute callbacks from `setTimeout()` and `setInterval()`
2. **Pending callbacks** — Execute I/O callbacks deferred from the previous loop
3. **Idle/Prepare** — Internal use only
4. **Poll** — Retrieve new I/O events; execute I/O callbacks. If no timers are scheduled, it will block here waiting for events.
5. **Check** — Execute `setImmediate()` callbacks
6. **Close callbacks** — Execute close event callbacks (e.g., `socket.on('close')`)

**Microtask queues (drain between EVERY phase):**
- `process.nextTick()` queue — drains first (highest priority)
- Promise microtask queue — drains second

**Key rules:**
- All synchronous code runs before any callbacks
- Between each phase transition, ALL microtasks are drained
- `nextTick` always runs before Promise callbacks
- Microtasks added during microtask processing run in the same cycle
- Recursive `nextTick` can starve I/O (event loop can't advance)

### Q4: What is the difference between `process.nextTick()`, `setImmediate()`, `setTimeout(fn, 0)`, and `Promise.then()`?

**Model Answer:**

| Mechanism | Type | When it runs | Priority |
|-----------|------|-------------|----------|
| `process.nextTick()` | Microtask | After current operation, before I/O | Highest (1st) |
| `Promise.then()` | Microtask | After nextTick queue drains | 2nd |
| `setTimeout(fn, 0)` | Macrotask | Timers phase (next loop or current if timer ready) | 3rd (varies) |
| `setImmediate()` | Macrotask | Check phase (after poll) | 4th (varies) |

**When to use each:**
- `process.nextTick()` — When you need to run something after the current operation completes but before any I/O. Example: emitting events after constructor returns.
- `Promise.then()` — Standard async flow control.
- `setImmediate()` — When you want to allow I/O to happen first. Safe for recursion.
- `setTimeout(fn, 0)` — When you want to defer to the next timers phase. Note: minimum delay is ~1ms, not truly 0.

**Important:** Inside an I/O callback, `setImmediate` always runs before `setTimeout(fn, 0)`. Outside I/O, the order is non-deterministic.

### Q5: What is the module wrapper function in Node.js?

**Model Answer:**

Before executing your code, Node.js wraps every file in a function:

```javascript
(function(exports, require, module, __filename, __dirname) {
    // Your code here
});
```

This wrapper provides:
- **`exports`** — shorthand reference to `module.exports`
- **`require`** — function to import other modules
- **`module`** — reference to the current module (has `.exports`, `.id`, `.filename`, etc.)
- **`__filename`** — absolute path to the current file
- **`__dirname`** — absolute path to the directory containing the file

**Why this matters:**
1. Variables declared with `var` are scoped to the module (not global) — unlike browsers
2. `this` at the top level is `module.exports`, not `global`
3. Each module gets its own scope — no polluting the global namespace
4. `__filename` and `__dirname` are per-module, not global constants

### Q6: How would you handle a CPU-intensive task in Node.js without blocking the event loop?

**Model Answer:**

Several strategies, depending on the use case:

1. **Worker Threads** (recommended for JS computation):
   ```javascript
   const { Worker } = require('worker_threads');
   const worker = new Worker('./heavy-task.js', { workerData: input });
   worker.on('message', (result) => { /* process result */ });
   ```

2. **Child Processes** (for external programs or full isolation):
   ```javascript
   const { fork } = require('child_process');
   const child = fork('./heavy-task.js');
   child.send(data);
   child.on('message', (result) => { /* process result */ });
   ```

3. **Break work into chunks** with `setImmediate()`:
   ```javascript
   function processChunk(data, index, callback) {
       // Process a small chunk
       if (index < data.length) {
           setImmediate(() => processChunk(data, index + 100, callback));
       } else {
           callback();
       }
   }
   ```

4. **Offload to a job queue** (Redis/BullMQ) and process asynchronously

5. **Use a C++ addon** via N-API for performance-critical code

The key principle: keep the event loop free for I/O callbacks.

### Q7: Compare `dependencies`, `devDependencies`, and `peerDependencies` in `package.json`.

**Model Answer:**

| Type | Installed in production? | Purpose | Example |
|------|------------------------|---------|--------|
| `dependencies` | Yes | Required at runtime | express, mongoose, jsonwebtoken |
| `devDependencies` | No (`npm install --production` skips them) | Development/build tools | jest, eslint, typescript, nodemon |
| `peerDependencies` | Depends (npm 7+ auto-installs) | Plugin compatibility — the *consumer* must provide this | react (for a React component library) |

**Common mistakes:**
- Putting `typescript` in `dependencies` (should be `devDependencies` — you only need it to compile, not run)
- Putting test frameworks in `dependencies`
- Not understanding `peerDependencies` — it means "I work with version X of this package, but I don't bundle it; you the consumer must have it installed"

In Docker production builds, use `npm ci --only=production` to skip devDependencies and reduce image size.
