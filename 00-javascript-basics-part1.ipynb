## Section 1: Variables & Declarations (MCQs 1–12)

**Key concepts:**
- **`var`**: Function-scoped, hoisted (initialized as `undefined`), allows redeclaration
- **`let`**: Block-scoped, hoisted but not initialized (Temporal Dead Zone), no redeclaration
- **`const`**: Block-scoped, must be initialized at declaration, prevents reassignment (not mutation of object/array contents)
- **Hoisting**: Declarations are moved to the top of their scope during compilation
- **Temporal Dead Zone (TDZ)**: Period between entering scope and declaration where `let`/`const` cannot be accessed

### MCQ 1
**What is the difference between `var`, `let`, and `const`?**

A) No difference  
B) `var` is function-scoped, `let` and `const` are block-scoped  
C) `let` is global, `var` is local  
D) `const` allows reassignment

<details>
<summary><b>Click to reveal answer</b></summary>

**Answer: B) `var` is function-scoped, `let` and `const` are block-scoped**

`var` is function-scoped and hoisted (initialized as undefined). `let` and `const` are block-scoped and hoisted but NOT initialized (temporal dead zone). `const` cannot be reassigned.
</details>

## Section 2: Data Types (MCQs 13–25)

**Key concepts:**
- **Primitives**: string, number, boolean, null, undefined, symbol, bigint (7 total)

## Section 4: Operators & Expressions (MCQs 39–50)

**Key concepts:**
- **`??` (nullish coalescing)**: Returns right operand only when left is `null` or `undefined`
- **`||`**: Returns first truthy or last operand (triggers on any falsy)
- **`?.` (optional chaining)**: Returns `undefined` instead of throwing when accessing null/undefined
- **`...` (spread)**: Expands iterables; in params it's "rest" and collects
- **Destructuring**: Unpack values from arrays/objects into variables
- **Short-circuit**: `&&` returns first falsy or last; `||` returns first truthy or last

### MCQ 39
**What is the nullish coalescing operator (`??`)?**

A) Same as `||`  
B) Returns right operand when left is `null` or `undefined` (NOT for other falsy values)  
C) Checks if a value is NaN  
D) A ternary shorthand

<details>
<summary><b>Click to reveal answer</b></summary>

**Answer: B) Returns right operand when left is `null` or `undefined` (NOT for other falsy values)**

`??` only checks for null/undefined, unlike `||` which triggers on any falsy value. `0 ?? 42` → 0 (0 is not null/undefined). `0 || 42` → 42 (0 is falsy).
</details>
