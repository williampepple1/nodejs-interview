## Section 5: Functions (MCQs 51-65)

This section covers function declarations vs expressions, arrow functions, pure functions, IIFEs, default parameters, rest parameters, callbacks, higher-order functions, currying, and generators.

### MCQ 51

What is the difference between function declarations and function expressions?

<details>
<summary>Answer</summary>

**B.** Function declarations are hoisted with their body (can be called before definition). Function expressions (`const fn = function(){})` are NOT hoisted — the variable is hoisted but undefined.

</details>

### MCQ 52

Output: `greet(); function greet() { console.log('Hello'); }`

- A) 'Hello'
- B) TypeError
- C) ReferenceError
- D) undefined

<details>
<summary>Answer</summary>

**A.** Function declarations are fully hoisted — you can call them before their definition in the code.

</details>

````javascript
greet();
function greet() { console.log('Hello'); }
````

### MCQ 53

Output: `greet(); var greet = function() { console.log('Hello'); };`

- A) 'Hello'
- B) TypeError: greet is not a function
- C) ReferenceError
- D) undefined

<details>
<summary>Answer</summary>

**B.** `var greet` is hoisted as undefined. Calling `undefined()` throws TypeError.

</details>

````javascript
try {
  greet();
  var greet = function() { console.log('Hello'); };
} catch (e) {
  console.log(e.name + ': ' + e.message);
}
````

### MCQ 54

What is an arrow function and how is it different from a regular function?

- A) Just shorter syntax
- B) Arrow functions don't have their own `this`, `arguments`, `super`, or `new.target`
- C) Arrow functions are faster
- D) Arrow functions can't return values

<details>
<summary>Answer</summary>

**B.** Arrow functions inherit `this` from the enclosing lexical scope. They can't be used as constructors (no `new`), don't have `arguments` object, and can't be used as methods.

</details>

**B.** Arrow function with implicit return (no curly braces = auto return). `2 + 3 = 5`.

</details>

````javascript
const add = (a, b) => a + b;
console.log(add(2, 3));
````

### MCQ 56

### MCQ 57

What is an IIFE (Immediately Invoked Function Expression)?

- D) A recursive function

<details>
<summary>Answer</summary>

**B.** IIFEs create a private scope, avoiding global namespace pollution. Common before `let`/`const` existed. Pattern: `(function() { ... })()` or `(() => { ... })()`.

### MCQ 63

### MCQ 64

**B.** Generators return an iterator. Call `.next()` to get the next `yield`ed value. They produce `{ value, done }` objects. Used for lazy evaluation, infinite sequences, and async control flow.

## Section 6: Scope (MCQs 66-75)

This section covers global, function, and block scope, the scope chain, lexical scope, variable shadowing, and strict mode.

### MCQ 66

What are the types of scope in JavaScript?

- A) Only global and local
- B) Global, function, and block scope
- C) Public and private
- D) Module and file

<details>
<summary>Answer</summary>

**B.** Global scope (accessible everywhere), function scope (`var` is function-scoped), block scope (`let`/`const` are block-scoped — within `{}`).

</details>

### MCQ 67

Output:
```javascript
for (var i = 0; i < 3; i++) { setTimeout(() => console.log(i), 0); }
```

- A) 0, 1, 2
- B) 3, 3, 3
- C) undefined
- D) 0, 0, 0

<details>
<summary>Answer</summary>

**B.** `var` is function-scoped — all three closures share the same `i`. By the time setTimeout fires, the loop is done and `i` is 3.

</details>

````javascript
for (var i = 0; i < 3; i++) { setTimeout(() => console.log(i), 0); }
````

### MCQ 68

Output:
```javascript
for (let i = 0; i < 3; i++) { setTimeout(() => console.log(i), 0); }
```

- A) 0, 1, 2
- B) 3, 3, 3
- C) undefined
- D) 0, 0, 0

<details>
<summary>Answer</summary>

**A.** `let` creates a new binding for each iteration. Each closure captures its own `i`.

</details>

````javascript
for (let i = 0; i < 3; i++) { setTimeout(() => console.log(i), 0); }
````

### MCQ 69

What is the scope chain?

- A) A linked list of DOM elements
- B) When a variable is not found in the current scope, JavaScript looks up through outer scopes until it finds it or reaches global scope
- C) A chain of function calls
- D) Module dependency chain

<details>
<summary>Answer</summary>

**B.** Inner scopes have access to outer scopes. The scope chain is established at function creation (lexical scope), not at call time.

</details>

### MCQ 70

What is lexical scope?

- A) Scope determined at runtime
- B) Scope determined by where the function is written (defined) in the source code
- C) Scope determined by where the function is called
- D) Scope that changes dynamically

<details>
<summary>Answer</summary>

**B.** JavaScript uses lexical (static) scoping — a function's scope is determined by where it's defined, NOT where it's called. This is why closures work.

</details>

## Section 7: Closures (MCQs 76-85)

This section covers closure definition, practical uses (data privacy, memoization), the module pattern, and common closure pitfalls.

### MCQ 76

What is a closure?

- A) Closing a browser tab
- B) A function that has access to variables from its outer (enclosing) function's scope, even after the outer function has returned
- C) A loop termination
- D) An error handler

<details>
<summary>Answer</summary>

**B.** Closures "close over" their lexical environment. The inner function retains a reference to the outer scope's variables.

</details>

### MCQ 77

Output:
```javascript
function makeCounter() { let count = 0; return () => ++count; }
const counter = makeCounter();
console.log(counter(), counter(), counter());
```

- A) 1, 1, 1
- B) 0, 1, 2
- C) 1, 2, 3
- D) undefined

<details>
<summary>Answer</summary>

**C.** The returned function closes over `count`. Each call increments the same `count` variable. `++count` pre-increments.

</details>

````javascript
function makeCounter() { let count = 0; return () => ++count; }
const counter = makeCounter();
console.log(counter(), counter(), counter());
````

### MCQ 78

What is a practical use of closures?

- A) Styling HTML
- B) Data privacy / encapsulation — creating private variables that can't be accessed from outside
- C) Importing modules
- D) Sorting arrays

<details>
<summary>Answer</summary>

**B.** Closures enable the module pattern, private state, function factories, and memoization. The inner function is the only way to access the outer function's variables.

</details>

### MCQ 79

What is memoization?

- A) Memorizing code
- B) Caching function results so that calls with the same arguments return the cached result instead of recomputing
- C) A memory leak
- D) A design pattern for databases

<details>
<summary>Answer</summary>

**B.** Uses closures to store a cache object: `function memoize(fn) { const cache = {}; return (...args) => { const key = JSON.stringify(args); return cache[key] || (cache[key] = fn(...args)); }; }`.

</details>

### MCQ 80

Output:
```javascript
function createFunctions() {
  const result = [];
  for (var i = 0; i < 3; i++) { result.push(function() { return i; }); }
  return result;
}
const fns = createFunctions();
console.log(fns[0](), fns[1](), fns[2]());
```

- A) 0, 1, 2
- B) 3, 3, 3
- C) undefined
- D) 2, 2, 2

<details>
<summary>Answer</summary>

**B.** Classic closure trap. `var i` is shared across all closures. When functions run, `i` is 3 (loop finished). Fix: use `let`, or create an IIFE `(function(j){ return function(){ return j; } })(i)`.

</details>

````javascript
function createFunctions() {
  const result = [];
  for (var i = 0; i < 3; i++) { result.push(function() { return i; }); }
  return result;
}
const fns = createFunctions();
console.log(fns[0](), fns[1](), fns[2]());
````

## Section 8: The `this` Keyword (MCQs 86-100)

This section covers how `this` behaves in different contexts: regular functions, methods, arrow functions, constructors, and how to use `call`, `apply`, and `bind`.

### MCQ 86

What does `this` refer to in a regular function call?

- A) The function itself
- B) The global object (or undefined in strict mode)
- C) The calling function
- D) null

<details>
<summary>Answer</summary>

**B.** In non-strict mode, `this` in a plain function call is the global object (`window` in browsers, `global` in Node.js). In strict mode, it's `undefined`.

</details>

### MCQ 87

What does `this` refer to inside a method?

- A) The global object
- B) The object the method is called on
- C) The function itself
- D) undefined

<details>
<summary>Answer</summary>

**B.** `obj.method()` → `this` is `obj`. `this` is determined by HOW the function is called, not where it's defined.

</details>

### MCQ 88

What does `this` refer to in an arrow function?

- A) The calling object
- B) The arrow function itself
- C) The `this` value from the enclosing lexical scope (it inherits `this`)
- D) Always undefined

<details>
<summary>Answer</summary>

**C.** Arrow functions don't have their own `this`. They inherit it from the surrounding code at definition time (lexical `this`).

</details>

</details>

````javascript
const obj = { name: 'Alice', greet: function() { return this.name; } };
console.log(obj.greet());
````

### MCQ 99

Output:
```javascript
const obj = {
  count: 0,
  increment() {
    setTimeout(function() { this.count++; console.log(this.count); }, 100);
  }
};
obj.increment();
```

- A) 1
- B) NaN
- C) undefined
- D) 0

<details>
<summary>Answer</summary>

**B.** Inside `setTimeout`'s regular function callback, `this` is the global object (not `obj`). `globalThis.count` is undefined, and `undefined + 1 = NaN`. Fix: use arrow function `() => { this.count++; }` which inherits `this` from `increment`.

</details>

````javascript
const obj = {
  count: 0,
  increment() {
    setTimeout(function() { this.count++; console.log(this.count); }, 100);
  }
};
obj.increment();
````

### MCQ 100

How do you fix the `this` problem in setTimeout callbacks?

- A) Only use `var`
- B) Use arrow functions (inherit `this`), `bind()`, or save reference `const self = this`
- C) Use `eval()`
- D) Use `new`

<details>
<summary>Answer</summary>

**B.** Three solutions: (1) Arrow function: `setTimeout(() => this.count++, 100)`, (2) Bind: `setTimeout(function() { this.count++ }.bind(this), 100)`, (3) Self reference: `const self = this; setTimeout(function() { self.count++ }, 100)`.

</details>
